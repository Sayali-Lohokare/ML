**Q - what prblm is a dict solving**

- Imagine a huge phone book. You want Rajeev’s number, but you don’t want to read every entry. You want to jump straight to the right page.

- List: checks entries one by one. The bigger it gets, the slower it is (O(n)).

- Dict: jumps straight to the answer, taking about the same tiny time however big it is (O(1) on average).

- The trick behind this is hashing. Python turns each key into a number that tells it exactly where to store and find the value. That is why a dict is called a hash table.



**Q - If set does the same task then why dict?**

- A set stores only items. A dict stores key → value pairs. Both use hashing to find things fast, but they solve different problems.

- In set, It can only tell you whether something exists. There is nothing attached to each item.

- In dict, It finds the key quickly and hands back the value stored with it. This is the actual phone book from your example: the name is the key, the number is the value.

- a set is for checking presence or removing duplicates. A dict is for looking up information linked to something. Same speed trick (hashing), different job.
    

**Q - What is Hashing**

- Hashing means converting any value into a number, using a built-in function called hash(). Think of it as a blender: you put anything in, and one number comes out.

**How a dict or set uses it**

- The hash number tells Python where to store the item in memory. Later, when you look it up, Python calculates the same number and jumps straight to that spot, with no searching. 

- That is why 'in' on a set was about 300 times faster than on your list.

**Rules**

- Same input gives the same hash, every time. hash("Rajeev") always gives the same number, so Python can find the item again where it stored it.

- Different inputs usually give different hashes. Sometimes two items get the same hash (a collision), but it is rare and Python handles it internally.

- Only immutable things can be hashed (numbers, strings, tuples). A list cannot, because it can change, and then its hash would change and Python would lose it.

- This is why dictionary keys and set items must be immutable. You can use a tuple as a key, but not a list.

**LIST/SET/DICT COMPREHENSIONS**

```python
L1=[i*i for i in range(1,11) if i%2==0] # here [] tells we want o/p in list form
print(L1)
```

**SET Comprehension**

```python
s1={i*i for i in range(1,11)}
print(s1) # o/p is not ordered also duplicate will be removed
```

```python
# Unique lengths
#use a set comp to build set of distinct name lengths

names={'ravi','shiv','ramu','sara','vidyutm','om'}
len_set={len(el) for el in names if len(el)%2==0} #here {} tells we want o/p in set type
print(len_set) #here duplicates are removed means same len names are not counted again
```

**DICT Comprehension**

```python
# dict comprehension

wrds=['hi','python','set']
lengths={len(w) for w in wrds}
leng={w:len(w) for w in wrds} # w: tells python that we want o/p in dict form
print(type(lengths))
print(type(wrds))
print(lengths)
print(leng)
print(type(leng))
```

```python
# price with tax - using .items() method

prices={"pen": 10, "book": 50, "bag": 200}
tax={item:price*1.18 for item,price in prices.items()}
print(tax)
```
